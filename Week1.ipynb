```python
#EXERCISE 1
h = 6.62607015e-34
c = 299792458
pi = 3.14159265359
hbar = h/(2 * pi)
e = 1.602176634e-19
me = 9.10938370e-31
ep0 = 8.85418781e-12
```

```python
#Use these to calculate Bohr Radius
a0 = (4*pi*ep0*hbar**2)/(me*e**2)
print(a0)
```

```python
#Make a variable for the n=1 level
e1 = -(e**2)/(8*pi*ep0*a0*1**2)
print(e1)
```

```python
#change to electronvolts by divinding by e
e1ev = e1/e
print(e1ev)
```

```python
#make a list with this as the only element to store the energy levels
ls = [e1ev]
print(ls)
```

```python
#append the next five values for n

for i in range(2,7):
    en = e1/(e*i**2)
    ls.append(en)

print(ls)

```

```python
#Cast this list into a set to remove any duplicates you may have created
s = set(ls)
print(s)
```

```python
#now cast back into a list
ls2 = list(s)
print(ls2)
```

```python
#sort the list
ls2.sort()
print(ls2)
```

```python
#sort the list in correct order
ls2.sort(reverse = True)
print(ls2)
```

```python
#Now calculate the first two wavelengths (i=3,4 f=2) of photons emitted in the visible Balmer series of transitions.
e4 = ls2[3]
e3 = ls2[2]
e2 = ls2[1]
lambda42 = (h*c)/(e4-e2)
lambda32 = (h*c)/(e3-e2)

#Convert back to SI Units and print
wl42 = lambda42/e
wl32 = lambda32/e

print(wl42)
print(wl32)
```
